```python
a, b = 0, 1
x = np.linspace(a, b, 1000)
def f(x, a, b):
    f = 2 * (x - a) / (b - a)**2
    return f
plt.plot(x, f(x, a, b))
plt.show()
```

```python
results = []
for _ in range(100000):

    x1 = np.random.uniform(a, b)
    y1 = np.random.uniform(0, 2)

    if (y1 < f(x1, a, b)):
        results.append(x1)
    else:
        continue

plt.hist(results, bins=10)
plt.show()
```
